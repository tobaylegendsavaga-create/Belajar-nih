UBOGY

UBOGY merupakan sebuah proyek mahasiswa dalam program PKKM yang berfokus pada pengembangan Umbrella Solar Technology, yaitu inovasi energi terbarukan berbasis panel surya yang dirancang hemat energi, praktis, dan memiliki unsur desain modern. Melalui akun Instagramnya, UBOGY mendokumentasikan proses kerja mulai dari persiapan alat, perakitan komponen, hingga pengenalan konsep teknologi dan manfaatnya kepada publik, serta membagikan konten edukatif mengenai proyek mereka. Selain mengerjakan aspek perangkat keras, tim ini juga terlibat dalam pemrograman dan pengelolaan sistem energi, menunjukkan bahwa UBOGY tidak hanya berperan sebagai tim teknis, tetapi juga sebagai kelompok riset dan pengembang teknologi hijau yang memadukan inovasi, edukasi, dan implementasi nyata dalam bidang energi berkelanjutan.

PESTRAP.ID

PESTRAP.ID merupakan proyek inovasi dari bidang Teknik Telekomunikasi yang berfokus pada pengembangan perangkap hama pintar berbasis IoT untuk meningkatkan efektivitas pengendalian hama, khususnya kumbang tanduk pada kelapa sawit. Melalui sensor, panel surya, dan sistem pemantauan otomatis, PESTRAP menyediakan solusi modern untuk deteksi dan penanganan hama tanpa intervensi manual. Selain itu, PESTRAP memiliki keterkaitan dengan layanan pengendalian hama profesional dari PT Pestrap Mitra Sukses, yang mencakup pest control, termite control, dan fumigasi. Secara keseluruhan, PESTRAP menggabungkan teknologi IoT, energi terbarukan, serta metode pengendalian hama konvensional untuk menciptakan solusi ramah lingkungan yang mendukung produktivitas pertanian dan keberlanjutan.

Cembong Official

Cembong Official adalah tim mahasiswa dari FTI ITERA yang terlibat dalam program PKKM 2025, dibimbing oleh dosen Zunanik Mufidah, S.TP., M.Si. Mereka memproduksi abon “Cembong” sebagai produk unggulan, mengedukasi audiens tentang proses pembuatannya melalui konten video reels dan postingan yang memperlihatkan tahap produksi. Melalui Instagram, mereka membagikan konten branding, dokumentasi kegiatan PKKM, serta memperkenalkan “Sobat Cembong” sebagai komunitas pendukung mereka, menunjukkan kombinasi antara semangat wirausaha mahasiswa dan pengembangan produk berbasis kampus.

Saibatin ITERA

Saibatin ITERA adalah tim riset dan pengembangan maritim dari Institut Teknologi Sumatera yang fokus pada kapal tanpa awak (unmanned vessel), dan telah berprestasi di kompetisi nasional seperti Etam Ship Competition, Semarak Mahasiswa Perkapalan, dan Kontes Kapal Indonesia. Mereka merancang inovasi kapal elektrik dengan material ramah lingkungan — misalnya lambung berlapis serat nanas, ubin dari serat kayu, dan kursi hasil daur ulang plastik — yang berhasil memenangkan Juara 1 Lomba Prototype dan Juara 2 Fun Race pada KKCTBN 2023. Tim ini terdiri dari sekitar 44 anggota, berkomitmen menggabungkan semangat mahasiswa dengan penelitian serius di bidang maritim, serta membawa nama baik ITERA di kancah nasional.

QuizIn

QuizinID adalah tim mahasiswa dari FTI ITERA yang mengembangkan aplikasi QUIZIN!, sebuah platform kuis interaktif untuk membuat proses belajar menjadi lebih seru, kolaboratif, dan menyenangkan. Mereka berbagi konten di Instagram yang menampilkan fitur-fitur aplikasi seperti game kuis, leaderboard, dan aspek edukasi, memperlihatkan kemajuan PKKM-FTI mereka. QuizinID memposisikan diri sebagai inovator edtech kampus yang menggabungkan aktivitas akademik dan interaktivitas digital untuk meningkatkan engagement belajar mahasiswa.


MECOZE

MECOZE (Mengolah Coconut Fiber Zero Waste) adalah tim mahasiswa dari FTI ITERA yang berpartisipasi di PKKM 2025, berfokus pada pemanfaatan limbah serabut kelapa menjadi produk bernilai tambah. Mereka membagikan konten di Instagram yang menampilkan proses pengolahan zero waste, edukasi tentang daur ulang serabut kelapa, dan dampak lingkungan yang positif melalui inovasi ramah lingkungan. MECOZE menggabungkan semangat wirausaha mahasiswa dengan solusi berkelanjutan untuk mengurangi limbah organik sambil menciptakan manfaat sosial dan ekonomi dari bahan alami yang biasanya dibuang.

ITERA HERO Agro Eduwisata 

ITERA HERO Agro Eduwisata merupakan inisiasi dari mahasiswa Prodi Pariwisata ITERA yang mengangkat Lembah Suhita sebagai destinasi wisata edukatif berkelanjutan, menggabungkan aspek konservasi, pertanian terpadu, dan pariwisata berbasis komunitas. Mereka menguatkan brand Lembah Suhita lewat FGD dengan berbagai pihak (akademisi, pelaku wisata, UMKM, pemerintah) dan mengembangkan paket wisata edukatif yang meliputi rumah lebah dan workshop konservasi. Dalam konten promosi digital, tim menekankan regenerasi lingkungan dan kearifan lokal untuk menarik minat wisatawan yang peduli pada ekowisata dan pendidikan lingkungan.

SIO SmartDoll

SIO SmartDoll adalah tim mahasiswa dari ITERA yang mengembangkan boneka pintar berbasis IoT sebagai alat terapi wicara untuk anak-anak dengan speech delay. Teknologi mereka memungkinkan interaksi real-time dan umpan balik, menjadikan terapi lebih menarik dan efektif. Inovasi ini diangkat dalam pemberitaan sebagai upaya kolaborasi antara teknik biomedis dan pemrograman, berpotensi memberikan solusi konkret untuk kebutuhan edukasi dan kesehatan anak

BioClove

BioClove adalah tim mahasiswa FTI ITERA yang ikut PKKM 2025 dan mengembangkan lilin aromaterapi ramah lingkungan berbahan dasar daun cengkeh. Mereka menyajikan konsep produk inovatif dengan wadah bambu yang alami dan estetis, serta mempromosikannya melalui konten branding dan teaser digital. BioClove memadukan kewirausahaan mahasiswa dengan keberlanjutan, mengangkat potensi lokal — cengkeh — sebagai sumber bahan baku hijau sekaligus menawarkan solusi relaksasi alami

Seduhrice PKKM FTI ITERA

Seduhrice PKKM FTI ITERA adalah tim mahasiswa dari Teknik Kimia yang mengembangkan nasi instan “SeduhRice” sebagai solusi praktis untuk konsumsi nasi berkualitas tanpa perlu memasak. Produk ini hadir dalam beberapa varian rasa khas, dipromosikan melalui konten Instagram berupa video reels dan foto yang menunjukkan proses penyajian serta kemasan praktis. Inisiatif ini didukung oleh ITERA dan FTI, dengan pembimbing dosen dari Himatekia ITERA, menunjukkan perpaduan kreativitas kewirausahaan dan inovasi pangan mahasiswa dalam program PKKM

Alena

Alena adalah tim PKKM FTI ITERA yang membuat notebook handmade estetik dan aromatik, menggunakan daun dan elemen alami sebagai bahan dasar. Mereka membagikan konten di Instagram yang menonjolkan proses kerajinan tangan, material daur ulang, dan estetika alami (“Art From Leaf and Nature”). Produk ini didukung oleh FTI ITERA, menunjukkan perpaduan antara kreativitas mahasiswa, keberlanjutan, dan wirausaha berbasis seni kerajinan alami.

MedCheck Station 

MedCheck Station adalah tim mahasiswa Teknik Biomedis dari FTI ITERA yang mengembangkan mesin cek kesehatan otomatis berbasis aplikasi desktop—alat ini mampu mengukur tanda vital seperti tekanan darah, detak jantung, saturasi oksigen, suhu tubuh, tinggi dan berat badan, kemudian menghitung Indeks Massa Tubuh (IMT) dan status gizi secara otomatis. Sistem MedCheck menggunakan antarmuka berbasis VB .NET dan database MySQL. 
Itera Repository. Mereka mempromosikan inovasi ini dalam program PKKM FTI 2025 sebagai “mesin pintar pengukur tanda vital” yang cepat, otomatis, dan andal. Hasil pengujian menunjukkan fungsionalitas berjalan dengan baik (100%) dan tingkat kenyamanan pengguna sangat tinggi menurut System Usability Scale.Itera RepositoryMedCheck Station memadukan semangat kewirausahaan mahasiswa dengan solusi kesehatan digital yang efisien dan praktis

STEM Board

STEM Board (PKKM FTI ITERA) adalah inisiatif mahasiswa-dosen yang mengembangkan media pembelajaran interaktif berbasis STEM untuk siswa SMA/SD, di mana melalui papan eksperimen siswa dapat langsung mempraktikkan konsep-konsep teknik dasar seperti rangkaian listrik seri dan paralel, sensor LDR, sensor sentuh, hingga simulasi sistem tenaga surya. Inovasi ini digunakan dalam kegiatan Pengabdian kepada Masyarakat (PkM) di SMA PGRI Tumijajar dan dirancang agar pembelajaran sains dan teknologi menjadi lebih aplikatif, kreatif, dan menarik, sekaligus meningkatkan keterampilan berpikir kritis dan pemahaman teknis generasi muda.

AgroWSil

AgroWSil (Agrowsil ID) adalah tim mahasiswa dari Teknologi Industri Pertanian ITERA yang mengembangkan pakan ruminansia berbasis limbah agroindustri dengan menjaga kualitas proses produksi dari bahan baku hingga hasil akhir. Mereka memasarkan produk “ransum pakan ternak” melalui Instagram yang menampilkan proses pembuatan serta keunggulan mutu pakan untuk performa ternak. Inisiatif ini mencerminkan perpaduan antara kewirausahaan agro-mahasiswa dan inovasi keberlanjutan, dengan fokus memaksimalkan nilai tambah limbah pertanian untuk mendukung produktivitas peternak lokal.

Ecoscent Diffuser

Ecoscent Diffuser adalah proyek mahasiswa yang mengembangkan reed diffuser / pengharum ruangan ramah lingkungan, kemungkinan berhubungan dengan limbah organik atau minyak esensial (berdasarkan konteks riset aromaterapi dan paten aromatik dari kampus). Mereka menyajikan konten edukatif dan promosi di Instagram tentang pembuatan wewangian alami, menjaga keberlanjutan melalui penggunaan bahan lokal atau daur ulang. Inovasi ini mencerminkan kombinasi kreatifitas wirausaha mahasiswa dengan kepedulian terhadap ekologi dan desain aroma yang estetis.
