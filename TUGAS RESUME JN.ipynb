PERCABANGAN

Percabangan (Conditional Statements)
Percabangan memungkinkan program membuat keputusan dan menjalankan blok kode tertentu berdasarkan kondisi yang terpenuhi.
Fungsi: Mengontrol alur eksekusi program.

Kata Kunci Utama:
    1. if: Mengevaluasi kondisi pertama. Blok kode di bawah if hanya dijalankan jika kondisi bernilai True.
    2. elif (singkatan dari "else if"): Mengevaluasi kondisi berikutnya jika kondisi if sebelumnya bernilai False.
    3. else: Blok kode default yang dijalankan jika semua kondisi if dan elif sebelumnya bernilai False.

```python
# Percabangan (Conditional Statements)

nilai = 85

if nilai >= 90:
    print("Selamat! Anda mendapatkan nilai A.")
elif nilai >= 80:
    print("Bagus, Anda mendapatkan nilai B.")
else:
    print("Anda perlu belajar lebih giat.")
```

PERULANGAN

Perulangan (Looping)
Perulangan digunakan untuk menjalankan blok kode yang sama berulang kali.
Ini sangat berguna untuk mengotomatisasi tugas yang repetitif.

Jenis Utama:
    1. for loop:Digunakan untuk mengiterasi (mengulangi) melalui urutan (seperti list, tuple, dictionary, string, atau range).
    Sering digunakan ketika Anda tahu berapa kali perulangan harus dilakukan.
    2. while loop: Digunakan untuk menjalankan blok kode selama kondisi tertentu bernilai True.
Perulangan akan berhenti ketika kondisinya menjadi False.

```python
# Perulangan (Looping) - Contoh 1: For Loop

print("Contoh For Loop:")
for i in range(5):
    print(f"Iterasi ke: {i}")


print("\nContoh While Loop:")
hitung = 0
while hitung < 3:
    print(f"Hitungan: {hitung}")
    hitung += 1
```

LIBRARY

Library (Pustaka)
Library (sering juga disebut package atau module) adalah koleksi kode Python yang sudah ditulis dan dapat digunakan kembali untuk memecahkan masalah tertentu tanpa harus menulis kode dari awal.

Fungsi: Menyediakan fungsionalitas tambahan, mempercepat pengembangan, dan menyederhanakan tugas kompleks.

Cara Menggunakan: Menggunakan kata kunci import untuk memuat library ke dalam program Anda.

Contoh Populer:
    1. math: Untuk operasi matematika kompleks (akar kuadrat, trigonometri, dll.).
    2. random: Untuk menghasilkan angka atau pilihan acak.
    3. pandas: Untuk analisis data dan manipulasi tabel.
    4. numpy: Untuk komputasi numerik, terutama dengan array.

```python
# Library (Pustaka) - Contoh: Menggunakan library 'math'

import math as m 

angka = 16


akar_kuadrat = m.sqrt(angka) 

print(f"Angka: {angka}")
print(f"Akar kuadrat dari {angka} adalah {akar_kuadrat}")
```
